# Reproducible quantum pairs benchmark
Run from `10-quantum-pairs-trading/research` or the project root using the project virtual environment. The data is synthetic; the simulator does not establish quantum advantage or live profitability.

```python
from pathlib import Path
import subprocess, sys
root = Path.cwd()
if root.name == "research":
    root = root.parent
assert (root / "configs/demo.yaml").exists()
def run(*args):
    subprocess.run([sys.executable, "-m", *args], cwd=root, check=True)
run("project.prepare_data", "--config", "configs/demo.yaml")
run("project.run", "--config", "configs/demo.yaml", "--solver", "all")
run("project.report", "--runs", "runs/", "--output", "reports/comparison.html")
```
