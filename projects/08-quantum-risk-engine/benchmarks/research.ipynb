# Quantum risk: a reproducible CPU experiment

Compare classical and quantum-simulated tail estimation on the same discrete loss distribution. No QPU is used. Open this notebook from the project root or benchmarks directory using the installed environment. Install Jupyter separately for notebook UI.


```python
from pathlib import Path
import json
import subprocess
import sys
root = Path.cwd()
if not (root / "configs/demo.yaml").exists():
    root = root.parent
assert (root / "configs/demo.yaml").exists()
```

## Verify an exact discrete fixture

For equal-mass losses 0, 1, 2, 3, the 50% left quantile is 1 and expected shortfall is 2.5.


```python
from risk.measures import empirical_risk
fixture = empirical_risk([0, 1, 2, 3], alpha=0.5)
assert fixture["var"] == 1 and fixture["cvar"] == 2.5
fixture
```

## Regenerate the experiment

Writes immutable seeded data, both solver runs, 40 matched-budget trials, HTML and a checksum manifest. Timings and UUIDs vary; model results and measurement seeds are reproducible.


```python
subprocess.run([sys.executable, "-m", "project.reproduce", "--config", "configs/demo.yaml"], cwd=root, check=True)
```

## Inspect the evidence

Quantum intervals are asymptotic profile-likelihood hulls; aggregate error intervals use Student-t intervals across five seeds. They do not prove asymptotic scaling.


```python
from IPython.display import HTML, display
display(HTML((root / "reports/reproduction/comparison.html").read_text()))
```

Configuration, environment, source and artifact hashes are recorded. Failed trials remain in the raw outputs.
