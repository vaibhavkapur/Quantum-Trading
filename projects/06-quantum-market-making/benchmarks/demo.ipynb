# Quantum market-making experiment
Run from this folder or the project root. Uses CPU-only CI settings and records raw trial outputs. No quantum advantage is assumed.


```python
from pathlib import Path
import os, subprocess, sys
root = Path.cwd()
if root.name == "benchmarks": root = root.parent
os.chdir(root)
subprocess.run([sys.executable, "-m", "project.prepare_data", "--config", "configs/ci.yaml"], check=True)
```

```python
subprocess.run([sys.executable, "-m", "project.run", "--config", "configs/ci.yaml", "--solver", "all"], check=True)
subprocess.run([sys.executable, "-m", "project.report", "--runs", "runs", "--output", "reports/comparison.html"], check=True)
```
